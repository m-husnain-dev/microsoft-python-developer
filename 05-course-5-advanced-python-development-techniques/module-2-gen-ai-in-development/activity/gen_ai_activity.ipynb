```python
from statistics import mean, median, multimode
from typing import Dict, List, Optional, Union


def calculate_descriptive_stats(
    numbers: List[Union[int, float]]
) -> Dict[str, Optional[Union[float, int]]]:
    """Calculates mean, median, and mode for a list of numbers.

    Handles empty lists and non-unique modes safely.
    """
    # Guard Clause: Empty List
    if not numbers:
        return {"mean": None, "median": None, "mode": None}

    # Calculations
    calc_mean = float(mean(numbers))
    calc_median = float(median(numbers))

    # Mode Calculation
    modes = multimode(numbers)
    if len(modes) == 1 and len(modes) < len(numbers):
        calc_mode = modes[0]
    else:
        calc_mode = None

    return {"mean": calc_mean, "median": calc_median, "mode": calc_mode}


# ==========================================
# VERIFICATION / TESTING
# ==========================================

# Test 1: Main Assignment Data
test_data = [1, 2, 3, 4, 4, 5, 5, 5]
res = calculate_descriptive_stats(test_data)

print("--- Assignment Test Case ---")
print(f"Mean   : {res['mean']}   (Expected: 3.625)")
print(f"Median : {res['median']}   (Expected: 4.0)")
print(f"Mode   : {res['mode']}     (Expected: 5)")

# Test 2: Edge Cases
print("\n--- Edge Case Tests ---")
print("No Mode Case [1, 2, 3] :", calculate_descriptive_stats([1, 2, 3]))
print("Empty List Case []     :", calculate_descriptive_stats([]))
```
