```python
from statistics import mean, median, multimode
from typing import Dict, List, Optional, Union


def calculate_descriptive_stats(
    numbers: List[Union[int, float]]
) -> Dict[str, Optional[Union[float, int]]]:
    """Calculates the mean, median, and unique mode for a list of numbers.

    This function processes a sequence of numerical values and returns key
    descriptive statistics. It includes robust edge-case handling for empty
    inputs and datasets without a distinct single mode.

    Args:
        numbers (List[Union[int, float]]): A list containing integers or
          floating-point numbers.

    Returns:
        Dict[str, Optional[Union[float, int]]]: A dictionary containing three
        keys:
            - 'mean' (Optional[float]): The arithmetic mean of the list, or
              None if empty.
            - 'median' (Optional[float]): The middle value of the list, or None
              if empty.
            - 'mode' (Optional[Union[int, float]]): The unique most frequent
              value. Returns None if empty or if no unique mode exists (e.g.,
              multimodal/uniform distributions).

    Raises:
        None: Safely catches empty inputs without raising ZeroDivisionError.

    Example:
        >>> calculate_descriptive_stats([1, 2, 3, 4, 4, 5, 5, 5])
        {'mean': 3.625, 'median': 4.0, 'mode': 5}
        >>> calculate_descriptive_stats([1, 2, 3])
        {'mean': 2.0, 'median': 2.0, 'mode': None}
    """
    # Guard Clause: Handle empty list
    if not numbers:
        return {"mean": None, "median": None, "mode": None}

    # Core Calculations
    calc_mean = float(mean(numbers))
    calc_median = float(median(numbers))

    # Mode Logic: Handle multi-modal or no unique mode cases
    modes = multimode(numbers)
    if len(modes) == 1 and len(modes) < len(numbers):
        calc_mode = modes[0]
    else:
        calc_mode = None  # No single distinct mode

    return {"mean": calc_mean, "median": calc_median, "mode": calc_mode}


# ==========================================
# VERIFICATION & OUTPUT
# ==========================================
test_data = [1, 2, 3, 4, 4, 5, 5, 5]
results = calculate_descriptive_stats(test_data)

print("--- Output Results ---")
print(f"Mean   : {results['mean']}")
print(f"Median : {results['median']}")
print(f"Mode   : {results['mode']}")
```
