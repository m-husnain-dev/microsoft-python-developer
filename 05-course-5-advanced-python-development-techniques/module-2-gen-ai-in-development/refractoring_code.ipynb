```python
import inspect
from typing import Any, Callable, Dict
import pandas as pd
```

```python
def log_execution(func: Callable) -> Callable:
    """Notebook cells ke execution time aur logs track karne ke liye decorator."""
    def wrapper(*args, **kwargs):
        print(f"⚡ [RUNNING]: {func.__name__}...")
        result = func(*args, **kwargs)
        print(f"✅ [FINISHED]: {func.__name__}")
        return result
    return wrapper
```

```python
class DataPipeline:
    """Notebook ke raw processing steps ko clean OOP class mein convert karna."""

    def __init__(self, data_path: str):
        self.data_path = data_path
        self.df: pd.DataFrame = None

    @log_execution
    def load_data(self) -> pd.DataFrame:
        """Data load karta hai."""
        self.df = pd.read_csv(self.data_path)
        return self.df

    @log_execution
    def clean_data(self) -> pd.DataFrame:
        """Missing values aur duplicates handle karta hai."""
        if self.df is not None:
            self.df = self.df.drop_duplicates().dropna()
        return self.df

    @log_execution
    def transform_features(self, target_column: str) -> Dict[str, Any]:
        """Notebook ke manual features ko dictionary/X, y splits mein refactor karna."""
        X = self.df.drop(columns=[target_column])
        y = self.df[target_column]
        return {"X": X, "y": y}
```

```python
if __name__ == "__main__":

    # Create a dummy data.csv file for demonstration
    dummy_data = {
        'feature_1': [10, 20, 15, 25, 30],
        'feature_2': [1, 2, 3, 4, 5],
        'target': [0, 1, 0, 1, 0]
    }
    dummy_df = pd.DataFrame(dummy_data)
    dummy_df.to_csv('data.csv', index=False)
    print("Created dummy 'data.csv' file.")

    pipeline = DataPipeline(data_path="data.csv")

    # Method chaining pattern
    pipeline.load_data()
    pipeline.clean_data()
    processed_data = pipeline.transform_features(target_column="target")

    print("\nRefactored Output Summary:")
    print(f"X shape: {processed_data['X'].shape}")
    print(f"y shape: {processed_data['y'].shape}")
```
