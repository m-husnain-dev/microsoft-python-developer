````python
import csv


class TriviaGameDriver:
    """Driver class to initialize, manage, and run the Trivia Game.

    Attributes:
        questions (list): A list of Question objects loaded from the CSV file.
        game (TriviaGame): An instance of the TriviaGame class managing game
          state.
    """

    def __init__(self, filename: str):
        """Initializes the TriviaGameDriver with questions loaded from a file
        and sets up the TriviaGame instance.

        Args:
            filename (str): The path to the CSV file containing trivia questions.
        """
        self.questions = self.load_questions_from_file(filename)
        self.game = TriviaGame(self.questions)

    def load_questions_from_file(self, filename: str) -> list:
        """Loads trivia questions from a CSV file, parses the contents, and

        instantiates Question objects.

        Args:
            filename (str): Path to the CSV file.

        Returns:
            list: A list of populated Question instances.
        """
        questions_list = []

        # Open file safely using context manager and parse CSV rows
        with open(filename, "r", encoding="utf-8") as csv_file:
            csv_reader = csv.reader(csv_file)
            next(csv_reader)  # Skip header row

            for row in csv_reader:
                question_text = row[0]
                answer_options = row[1].split("|")
                correct_answer = row[2]

                # Append newly instantiated Question object
                questions_list.append(
                    Question(question_text, answer_options, correct_answer)
                )

        return questions_list

    def start_game(self):
        """Starts and controls the main interactive loop for the trivia game."""
        while True:
            self.game.play_round()

            # Prompt user for continuing or exiting the game loop
            play_again = input("Play another round? (y/n): ")
            if play_again.lower() != "y":
                break

        print(f"\nFinal score: {self.game.score}")
````
