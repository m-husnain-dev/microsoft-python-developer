```python
# Cell 2: Test sample Python library file to generate (math_utils.py)
%%writefile math_utils.py
"""
Math Utils Module
=================
this module basic descriptive statistics and perform arithmetic operations .
"""

from statistics import mean, median, multimode
from typing import List, Dict, Union, Optional

def calculate_stats(numbers: List[Union[int, float]]) -> Dict[str, Optional[Union[float, int]]]:
    """List ke liye mean, median, aur unique mode calculate karta hai.

    Args:
        numbers (List[Union[int, float]]): Numerical values list.

    Returns:
        Dict[str, Optional[Union[float, int]]]: Stats dictionary.
    """
    if not numbers:
        return {"mean": None, "median": None, "mode": None}

    modes = multimode(numbers)
    mode_val = modes[0] if len(modes) == 1 and len(modes) < len(numbers) else None

    return {
        "mean": float(mean(numbers)),
        "median": float(median(numbers)),
        "mode": mode_val
    }
```
