```python
# Cell 1: Bad Practice - Messy, Non-Pythonic, & Nested Code
def process_orders_bad(orders):
    result = []
    for i in range(len(orders)):
        if orders[i]["status"] == "completed":
            total = orders[i]["price"] * orders[i]["quantity"]
            if orders[i]["discount"] > 0:
                total = total - (total * (orders[i]["discount"] / 100))
            result.append(
                {"id": orders[i]["id"], "final_price": round(total, 2)}
            )
    return result


# Sample Test Data
raw_orders = [
    {"id": 101, "price": 50.0, "quantity": 2, "discount": 10, "status": "completed"},
    {"id": 102, "price": 30.0, "quantity": 1, "discount": 0, "status": "pending"},
    {"id": 103, "price": 100.0, "quantity": 3, "discount": 15, "status": "completed"},
]

print("Bad Code Output:", process_orders_bad(raw_orders))
```

```python
# Cell 2: Step 1 - Calculation logic ko extract karna aur Type Hints add karna
from typing import Dict, Any


def calculate_discounted_total(price: float, quantity: int, discount_percent: float) -> float:
    """Order ka final price discount subtract karke calculate karta hai."""
    subtotal = price * quantity
    discount_amount = subtotal * (discount_percent / 100)
    return round(subtotal - discount_amount, 2)
```

```python
# Cell 3: Step 2 - List Comprehension aur Guard Clauses ke sath final refactored code
from typing import List


def process_order_item(order: Dict[str, Any]) -> Dict[str, Any]:
    """Single order object ko process karke standard output format karta hai."""
    final_price = calculate_discounted_total(
        price=order.get("price", 0.0),
        quantity=order.get("quantity", 0),
        discount_percent=order.get("discount", 0.0),
    )
    return {"id": order["id"], "final_price": final_price}


def process_orders_clean(orders: List[Dict[str, Any]]) -> List[Dict[str, Any]]:
    """Clean, readable aur maintainable refactored main function."""
    return [
        process_order_item(order)
        for order in orders
        if order.get("status") == "completed"
    ]
```

```python
# Cell 4: Verify karna ke Refactoring ke baad Output exact same hai ya nahi
bad_output = process_orders_bad(raw_orders)
clean_output = process_orders_clean(raw_orders)

print("Refactored Code Output:", clean_output)
print("Outputs Match Perfectly?:", bad_output == clean_output)
```
