```python
%%writefile test_bank_account.py
import pytest

# =====================================================================
# 1. APPLICATION CODE (Green Phase Implementation)
# =====================================================================


class BankAccount:
    """Class representing a bank account with deposit and withdraw functionality."""

    def __init__(self, owner: str, initial_balance: float = 0.0):
        if not owner or not isinstance(owner, str):
            raise ValueError("Owner name must be a non-empty string.")
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")

        self.owner = owner
        self.balance = initial_balance

    def deposit(self, amount: float) -> float:
        if amount <= 0:
            raise ValueError("Deposit amount must be greater than zero.")
        self.balance += amount
        return self.balance

    def withdraw(self, amount: float) -> float:
        if amount <= 0:
            raise ValueError("Withdrawal amount must be greater than zero.")
        if amount > self.balance:
            raise ValueError("Insufficient funds for this withdrawal.")

        self.balance -= amount
        return self.balance


# =====================================================================
# 2. TDD UNIT TESTS (Success + Failure / Edge Cases)
# =====================================================================


# --- Success Cases (Happy Paths) ---


def test_account_creation_success():
    account = BankAccount("Ali", 100.0)
    assert account.owner == "Ali"
    assert account.balance == 100.0


def test_deposit_valid_amount():
    account = BankAccount("Ali", 100.0)
    new_balance = account.deposit(50.0)
    assert new_balance == 150.0
    assert account.balance == 150.0


def test_withdraw_valid_amount():
    account = BankAccount("Ali", 100.0)
    new_balance = account.withdraw(40.0)
    assert new_balance == 60.0
    assert account.balance == 60.0


# --- Failure & Edge Cases ---


def test_create_account_negative_balance_raises_error():
    with pytest.raises(ValueError, match="Initial balance cannot be negative."):
        BankAccount("Ali", -50.0)


def test_create_account_invalid_owner_raises_error():
    with pytest.raises(ValueError, match="Owner name must be a non-empty string."):
        BankAccount("")


def test_deposit_zero_or_negative_amount_raises_error():
    account = BankAccount("Ali", 100.0)
    with pytest.raises(
        ValueError, match="Deposit amount must be greater than zero."
    ):
        account.deposit(0)

    with pytest.raises(
        ValueError, match="Deposit amount must be greater than zero."
    ):
        account.deposit(-20)


def test_withdraw_insufficient_funds_raises_error():
    account = BankAccount("Ali", 50.0)
    with pytest.raises(
        ValueError, match="Insufficient funds for this withdrawal."
    ):
        account.withdraw(100.0)


def test_withdraw_negative_amount_raises_error():
    account = BankAccount("Ali", 100.0)
    with pytest.raises(
        ValueError, match="Withdrawal amount must be greater than zero."
    ):
        account.withdraw(-10.0)
```
